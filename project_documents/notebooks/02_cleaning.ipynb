```python
import psycopg2
import pandas as pd
conn = psycopg2.connect(dbname="accio_retailmart_27", user="postgres", password="postgres", host="localhost", port="5432")
print("Connected for cleaning validation")
```

### 1. Delivery Sequence Verification

```python
query_seq = """
SELECT count(*) as invalid_shipments
FROM sales.shipments
WHERE delivered_date < shipped_date;
"""
pd.read_sql(query_seq, conn)
```

### 2. Financial Boundary Checks

```python
query_fin = """
SELECT 
  count(CASE WHEN net_total < 0 THEN 1 END) as negative_orders,
  count(CASE WHEN discount_amount > gross_total THEN 1 END) as excessive_discount
FROM sales.orders;
"""
pd.read_sql(query_fin, conn)
```
