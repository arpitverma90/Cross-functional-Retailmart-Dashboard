# RetailMart V3 - Notebook 01: Data Profiling
Comprehensive structural and statistical profiling of raw datasets across 16 schemas and 55 tables.

```python
import os
import psycopg2
import pandas as pd
from dotenv import load_dotenv
load_dotenv()
conn = psycopg2.connect(dbname="accio_retailmart_27", user="postgres", password="postgres", host="localhost", port="5432")
print("Connected to accio_retailmart_27")
```

### 1. Table Counts and Grains Summary

```python
query = """
SELECT table_schema, count(*) as tables, sum(n_live_tup) as total_estimated_rows
FROM pg_stat_user_tables
WHERE table_schema NOT IN ('analytics')
GROUP BY table_schema ORDER BY table_schema;
"""
df_summary = pd.read_sql(query, conn)
df_summary
```

### 2. Commercial Orders & Revenue Profiling

```python
query_orders = """
SELECT order_status, count(*) as order_count, sum(net_total) as total_net_revenue, round(avg(net_total), 2) as aov
FROM sales.orders
GROUP BY order_status ORDER BY total_net_revenue DESC;
"""
df_orders = pd.read_sql(query_orders, conn)
df_orders
```
