### EDA Question 1: Monthly Net Sales Velocity

```python
q1 = """
SELECT 
    to_char(date_trunc('month', order_date), 'YYYY-MM') as yr_month,
    count(*) as delivered_orders,
    round(sum(net_total) / 10000000.0, 2) as net_revenue_crores,
    round(avg(net_total), 2) as aov
FROM sales.orders
WHERE order_status = 'Delivered'
GROUP BY date_trunc('month', order_date)
ORDER BY yr_month;
"""
df_monthly = pd.read_sql(q1, conn)
df_monthly
```
